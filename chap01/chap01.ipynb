## Свойства преобразования Фурье

#### 1. Линейность: если $$ s(t) = \alpha f(t) + \beta g(t), $$ то $$ \dot{S}(t) = \alpha \dot{F}(\omega) + \beta\dot{G}(\omega)$$ Спектр суммы равен сумме спектров

#### 2. Задержка: если $$ s(t) = f(t - \tau), $$ то $$ \dot{S}(t) = \int_{-\infty}^{\infty}{f(t - \tau)e^{-j\omega t}dt} = \int_{-\infty}^{\infty}{f(t - \tau)e^{-j\omega (t - \tau)}d(t - \tau)e^{-j\omega \tau}dt = \dot{F}(\omega)e^{-j\omega \tau}}$$ Спектр исходного сигнала уножается на комплексную экспоненту, амплитудный спектр не меняется, фазовые спектр приобретает дополнительное слагаемое $-\omega\tau$

#### 3. Изменение масштаба оси времени: если $$ s(t) = f(at),$$ то $$ \dot{S}(t) = \int_{-\infty}^{\infty}{f(at)e^{-j\omega t}dt} = \frac{1}{a} \int_{-\infty}^{\infty}{f(at)e^{-j\frac{\omega}{a}at}d(at)} = \frac{1}{a} \dot{F}(\frac{\omega}{a})$$ Изменение длительности сигнала приводит к изменению ширины спектра в противоположную сторону в сочетании с увеличением или уменьшением уровня спектральных составляющих.

#### 4. Дифференцирование сигнала: если $$ s(t) = \frac{df(t)}{dt} = \lim_{\varepsilon \to 0}{\frac{f(t + \varepsilon) - f(t)}{\varepsilon}}$$, то $$ \dot{S}(\omega) = \int_{-\infty}^{\infty}\lim_{\varepsilon \to 0}{\frac{f(t + \varepsilon) - f(t)}{\varepsilon}}e^{-j\omega t}dt = \lim_{\varepsilon \to 0}{\frac{\dot{F}(\omega)e^{j\omega \varepsilon} - \dot{F}(\omega)}{\varepsilon}} = \dot{F}(\omega)\lim_{\varepsilon \to 0}{\frac{e^{j\omega\varepsilon} - 1}{\varepsilon}} = j\omega\dot{F}(\omega)$$ Спектр производной получается путем умножения исходного сигнала на $j\omega$. Таким образом, при дифференцировании низкие частоты ослабляются, а высокие усиливаются. Фазовый спектр сдвигается на 90 градусов для положительных частот и на -90 для отрицательных.

#### 5. Интегрирование сигнала: $$ \dot{S}(\omega) = \frac{\dot{F}(\omega)}{j\omega} + \pi\dot{F}(0)\delta(\omega) $$

#### 6. Спектр свертки сигналов: $$ \dot{S}(\omega) = \int_{-\infty}^{\infty}{\int_{-\infty}^{\infty}{f(t^\prime)g(t - t^\prime)dt^\prime e^{-j\omega t}dt}} = \int_{-\infty}^{\infty}{f(t^\prime)e^{-j\omega t}}\int_{-\infty}^{\infty}{g(t - t^\prime)e^{-j\omega t}d(t - t^\prime)dt^\prime} = \dot{F}(\omega)\dot{G}(\omega)$$ Спектр свертки равен произведению спектров

#### 7. Спектр произведения: $$ \dot{S}(\omega) = \int_{-\infty}^{\infty}{f(t)g(t)e^{-j\omega t}dt} = \int_{-\infty}^{\infty}{\left(\frac{1}{2\pi}\int_{-\infty}^{\infty}{\dot{F}(\omega^\prime)e^{-j\omega^\prime t}d\omega^\prime}\right)g(t)e^{-j\omega t}dt} = \frac{1}{2\pi}\int_{-\infty}^{\infty}{\dot{F}(\omega^\prime)\int_{-\infty}^{\infty}{g(t)e^{-j(\omega - \omega^\prime)t}dtd\omega^\prime}} = \frac{1}{2\pi} \int_{-\infty}^{\infty}{\dot{F}(\omega^\prime)\dot{G}(\omega - \omega^\prime)d\omega^\prime}$$ Спектр произедения представляет собой свертку спектров

## Упражнение 1.2

````python
from thinkdsp.thinkdsp import *
````

````python
wave = read_wave('4.wav')
wave.make_audio()
````

````python
wave.plot()
````

````python
segment = wave.segment(start=0.63, duration=0.3)
segment.make_audio()
````

````python
segment.plot()
````

````python
segment.segment(start=0.64, duration=0.05).plot()
````

````python
spectrum = segment.make_spectrum()
spectrum.plot(high=4000)
````

Составляющих в сигнале не слишком много, так как звук практический чистый, связано это с тем, что во фрагменте звучит арфа, ноты извлекаются по одной щипками. Посмотрим на пики и их частоты с помощью ```peaks()```

````python
spectrum.peaks()[:30]
````

Основная частота, по совместительству доминирующая - 293.33 Гц, что близко к 293.66 - нота Ре первой октавы.

Тембр звука практически чистейший, что можно наблюдать и в спектре, гармоники выражены слабо.

````python
spectrum.low_pass(600)
````

````python
spectrum.make_wave().make_audio()
````

````python
spectrum.plot(high=1000)
````

Однако же, при обрезании высоких частот звук стал заметно глуше

````python
signal_sin = (SinSignal(freq=200, amp=1.5) +
          SinSignal(freq=400, amp=0.4) +
          SinSignal(freq=600, amp=0.625))
signal_sin.plot()
````

````python
wave2 = signal_sin.make_wave(duration=2)
wave2.apodize()
````

````python
wave2.make_audio()
````

````python
spectrum = wave2.make_spectrum()
spectrum.plot(high=1000)
````

````python
signal_sin += SinSignal(freq=650)
signal_sin.make_wave().make_audio()
````

Если добавить сигнал, частота которого не кратка основной (200), то услышим эту составляющую отчетливо.

Попробуем проделать все то же самое с косинусом

````python
signal_cos = (CosSignal(freq=100, amp=1) +
              CosSignal(freq=200, amp=1.5) +
              CosSignal(freq=300, amp=0.7))
signal_cos.plot()
````

````python
wave3 = signal_cos.make_wave(duration=2)
wave3.apodize()
````

````python
wave3.make_audio()
````

````python
spectrum = wave3.make_spectrum()
spectrum.plot(high=1000)
````

## Упражнение 1.4

````python
def stretch(wave, alpha):
    wave.ts /= alpha
    wave.framerate *= alpha
````

````python
wave4 = read_wave('The-Elder-Scrolls-III_-Morrowind-Peaceful-Waters.wav')
wave4.normalize()
wave4 = wave4.segment(start=0, duration=30)
wave4.make_audio()
````

````python
stretch(wave4, 2)
wave4.make_audio()
````

````python
wave4.plot()
````
